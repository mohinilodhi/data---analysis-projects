```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Department": ["IT", "HR", "Finance", "IT"]
})

df2 = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Salary": [50000, 60000, 70000, 55000]
})



df = pd.merge(df1, df2, on="Name")
print(df)
```

```python
result = df.groupby("Department")["Salary"].max()
print(result)
```

```python
result = df.loc[df.groupby("Department")["Salary"].idxmax()]
print(result)
```

```python
df["Sales"] = [200000, 150000, 300000, 250000]
```

```python
df["Efficiency"] = df["Sales"] / df["Salary"]
print(df)
```

```python
df["Efficiency"]
```

```python
result = df.groupby("Department")["Efficiency"].mean()
print(result)
```

```python
best = df.loc[df["Efficiency"].idxmax()]
print(best)
```

```python
bad = df.loc[df["Efficiency"].idxmin()]
print(bad)
```

```python
df.loc[df.groupby("Department")["Efficiency"].idxmin()]
```

```python
df.loc[df.groupby("Department")["Efficiency"].idxmax()]
```

```python
df["Efficiency"] = [0,5,28,54]
```

```python
df["Performance"] = df["Efficiency"].apply(
    lambda x: "High" if x > 5 else ("Medium" if x >= 3 else "Low")
)

print(df[["Name", "Efficiency", "Performance"]])
```
