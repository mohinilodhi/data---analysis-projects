```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Department": ["IT", "HR", "Finance", "IT"],
    "Salary": [50000, 60000, 70000, 55000],
    "Sales": [200, 150, 300, 250]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.isnull()
```

```python
df.dropna()
```

```python
df.duplicated()
```

```python
df.drop_duplicates()
```

```python
df.pivot_table(values="Salary",index="Department",aggfunc ="mean")
```

```python
df.pivot_table(values="Salary", index="Department", aggfunc="sum")
```

```python
df.pivot_table(values="Salary", index="Department", aggfunc=["max","mean","min"])
```

```python
df.pivot_table(values=["Salary","Sales"], index="Department",aggfunc="mean")
```

```python
df.pivot_table(
    values="Salary",
    index="Department",
    columns="Name",
    aggfunc="sum",
    fill_value=0
)
```

```python
df["Salary"] = df["Salary"].apply(lambda x: x *2)
print(df)
```

```python
df["Sales"] = df["Sales"].apply(lambda x: x *4)
print(df)
```

```python
df["Status"] = df["Sales"].apply(
    lambda x: "Good" if x> 100 else "average"
)
print(df)
```

apply with multiple columns


```python
df["Bonus"] = df.apply(
    lambda row:
    row["Salary"] * 0.20 if row["Sales"] > 200
    else row["Salary"] * 0.10,
    axis=1
)
print(df)
```

```python
df["Efficiency"] = df["Sales"] / df["Salary"]
print(df)
```

```python
best = df.loc[df["Efficiency"].idxmax()]
print(best)
```

```python
bad = df.loc[df["Efficiency"].idxmin()]
print(bad)
```
